```python
import pandas as pd

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom'],
    'score': [85, 92, 78, 88, 95],
    'age': [20, 21, 19, 22, 20]
}

df = pd.DataFrame(data)
print(df)
```

```python
#정렬
df.sort_values('score',ascending=False)
```

```python
df.sort_values('name')
```

```python
#여러 조건 필터
df[(df['score']>=80) & (df['age']>=20)]
```

```python
df[(df['score']>=90) & (df['age']<20)]
```

```python
#column 추가
df['pass'] = df['score']>=80
print(df)
```

```python
#간단 분석
df['age'].mean()
```

```python
#score 기준 내림차순 정렬
print(df.sort_values('score', ascending = False))
```

```python
#age가 20 이상인 사람만 출력
print(df[df['age']>=20])
```

```python
#score ≥ 80 AND age ≥ 20인 사람 출력
df[(df['score']>=80) & (df['age']>=20)]
```

```python
#score ≥ 90인 사람의 이름만 출력
df[(df['score']>=90)]['name']
```

```python
#pass 컬럼 추가 (80 이상 = True)
df['pass'] = df['score']>=80
print(df)
```

```python
#pass가 True인 사람들의 평균 score
df[df['pass']==True]['score'].mean()
```
