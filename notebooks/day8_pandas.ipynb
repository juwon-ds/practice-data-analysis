```python
import pandas as pd

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom', 'John'],
    'age': [20, 21, 19, 22, 20, 23],
    'score': [85, 90, 65, 88, 95, 70],
    'gender': ['F', 'F', 'M', 'F', 'M', 'M']
}

df = pd.DataFrame(data)
df
```

```python
# 조건 필터링

# 점수가 80 이상이고, 성별이 F인 사람
df[(df['score']>=80) & (df['gender']=='F')]
```

```python
# 정렬

# 점수 높은 순으로 정렬
df.sort_values('score', ascending=False)
```

```python
df.query('age >= 20 and score >= 80')
```

```python
df.query('score >= 85 and gender == "M"')
```

```python
# groupby

# 성별별로 평균 점수, 최대 점수 한 번에 구하기

df.groupby('gender')['score'].agg(['mean', 'max'])
```

```python
df.query('score >= 80').groupby('gender')['score'].mean()
```
