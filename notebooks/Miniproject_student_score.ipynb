```python
import pandas as pd

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom', 'John', 'Emma', 'Chris', 'Lina', 'David'],
    'gender': ['F', 'F', 'M', 'F', 'M', 'M', 'F', 'M', 'F', 'M'],
    'study_time': [2, 4, 1, 3, 5, 2, 4, 3, 2, 5],
    'score': [75, 90, 65, 85, 95, 70, 88, 80, 78, 92]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.info()
```

```python
df.sort_values('study_time')
```

```python
df.isnull().sum()
```

Question :

Is there a difference in score by gender?

Is there a relationship between study time and score?

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))

# graph 1
plt.subplot(1,3,1)
df.groupby('gender')['score'].mean().plot(kind='bar')
plt.title('Average Score by Gender')

# graph 2
plt.subplot(1,3,2)
plt.scatter(df['study_time'], df['score'])
plt.title('Study time vs Score')

# graph 3
plt.subplot(1,3,3)
df['score'].plot(kind='hist')
plt.title('Distribution of Score')

plt.tight_layout()
plt.show()
```

Analysis

- Bar chart shows the average score by gender.

- Scatter plot shows the relationship between study time and score.

- Histogram shows the distribution of score.

Conclusion

- There is no clear difference by gender.

- There is a relation between study time and score. People who study more can get the higher score.
