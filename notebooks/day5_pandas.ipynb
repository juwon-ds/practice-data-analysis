```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom'],
    'score': [85, 92, 78, 88, 95],
    'age': [20, 21, 19, 22, 20],
    'gender': ['F', 'F', 'M', 'F', 'M']
}

df = pd.DataFrame(data)
df.head()
```

```python
# bar

df.groupby('gender')['score'].mean().plot(kind='bar')
plt.show()
```

```python
# histogram

df['score'].plot(kind='hist', bins=5)
plt.show()
```

```python
import numpy as np

scores = np.concatenate([
    np.random.normal(60, 5, 50),
    np.random.normal(90, 5, 50)
])

df = pd.DataFrame({'score': scores})
df['score'].plot(kind='hist', bins=10)
plt.show()

# hist는 distribution을 보는 것! 정확한 값 X!!
```

```python
# scatter

plt.scatter(df['age'], df['score'])
plt.show()
```

```python
plt.title('Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Score')
```

Quiz

```python
# 성별 평균 점수 막대그래프 그리기

df.groupby('gender')['score'].mean().plot(kind='bar')
plt.title('Average Score by Gender')
plt.show()

# The average score by gender of Female is the higer than Male.
```

```python
# 나이 분포 히스토그램

df['age'].plot(kind='hist')
plt.title('Score Distribution')
plt.show()

# Distribution of the score clusters between 19.5 and 20.5.
```

```python
# 나이 VS 점수 산점도

plt.scatter(df['age'], df['score'])
plt.title('Age vs Score')
plt.xlabel('Age')
plt.ylabel('Score')
plt.show()

# There is no clear correlation between age and score
```
