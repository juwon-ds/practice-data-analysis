```python
# CSV 데이터 사용

import pandas as pd

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom'],
    'score': [85, None, 78, 88, None],
    'age': [20, 21, None, 22, 20]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 결측치 확인

df.isnull()
```

```python
df.isnull().sum()
```

```python
# 결측치 제거

df.dropna()
```

```python
df.dropna(subset=['score'])
```

```python
# 결측치 채우기

df['score'].fillna(df['score'].mean())
```

```python
df['score'] = df['score'].fillna(df['score'].mean())
```

```python
print(df)
```

```python
# 실전에서의 사용 예

df.head()
df.isnull().sum()
df['score'] = df['score'].fillna(df['score'].mean())
df.groupby('name')['score'].mean()
```

Quiz

```python
# 각 컬럼의 결측치 개수 구하기

df.isnull().sum()
```

```python
# score 결측치 있는 행 삭제

df.dropna(subset='age')
```

```python
# score 결측치를 평균으로 채우기

df.fillna(df['score'].mean())
```

```python
# age 결측치를 평균으로 채우고 성별 평균 나이 구하기

df['gender'] = ['F', 'F', 'M', 'F', 'M']
df.fillna(df['age'].mean())
df.groupby('gender')['age'].mean()
```

```python
# score 결측치 처리 후 성별 평균 점수 구하기

df.dropna(subset = 'score')
df.groupby('gender')['score'].mean()
```
