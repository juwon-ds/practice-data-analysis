```python
import pandas as pd

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom', 'John'],
    'score': [85, 92, 78, 88, 95, 70],
    'age': [20, 21, 19, 22, 20, 21],
    'gender': ['F', 'F', 'M', 'F', 'M', 'M']
}

df = pd.DataFrame(data)
print(df)
```

df.groupby('기준')['값'].함수()

--> 기준 여러 개인 경우 (['기준'])

--> 값 여러 개인 경우 [['값']]

```python
df.groupby('gender')['score'].mean()
```

```python
df.groupby('gender')['score'].max()
```

```python
df.groupby('gender')['name'].count()
```

Quiz

```python
#성별 평균 나이 구하기
df.groupby('gender')['age'].mean()
```

```python
#성별 최고 점수 구하기
df.groupby('gender')['score'].max()
```

```python
#성별 인원 수 구하기
df.groupby('gender')['name'].count()
```

```python
#성별 평균 점수가 더 높은 성별 찾기
df.groupby('gender')['score'].mean().idxmax()

#max() = 값
#idxmax() = 위치(이름)
```

```python
#성별 평균 점수 + 평균 나이 동시에 보기
df.groupby('gender')[['score','age']].mean()

#여러 컬럼은 이중 대괄호 [[]] !!
```
