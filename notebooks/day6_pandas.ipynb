```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'name': ['Juwon', 'Minji', 'Alex', 'Sara', 'Tom', 'John'],
    'score': [85, 92, 78, 88, 95, 70],
    'age': [20, 21, 19, 22, 20, 21],
    'gender': ['F', 'F', 'M', 'F', 'M', 'M']
}

df = pd.DataFrame(data)
df.head()
```

```python
plt.figure(figsize=(10,4)) # 종이 크기 정하기->가로 10, 세로4

# 1번 그래프
plt.subplot(1,2,1) # plt.subplot(행,열,위치)
df.groupby('gender')['score'].mean().plot(kind='bar')
plt.title('Average Score by Gender')

# 2번 그래프
plt.subplot(1,2,2)
plt.scatter(df['age'], df['score'])
plt.title('Age vs Score')

plt.tight_layout() # 그래프 간격 자동 정리
plt.show()
```



*   Analysis

1.   Question (문제 정의)
2.   Analsis (그래프 / 분석)
3.   Conclusion (그래프 보고 한 줄 정리)





Analysis

1. Question

*   Is there a difference in scores by gender?
*   Is there a relationship between age and score?

2. Analysis

*   Bar chart: average score by gender
*   Scatter plot: age vs score

3. Conclusion
There is a difference in average

*   There is a difference in average scores by gender.
*   There is no clear relationship between age and score.

```python
# subplot으로 그래프 만들기

plt.figure(figsize=(10,4))
plt.subplot(1,3,1)
df.groupby('gender')['score'].mean().plot(kind='bar')
plt.title('Average score by gender')

plt.subplot(1,3,2)
plt.scatter(df['age'], df['score'])
plt.title('Age vs Score')

plt.subplot(1,3,3)
df.groupby('gender')['age'].mean().plot(kind='bar')
plt.title('Average age by gender')

plt.tight_layout()
plt.show()
```

Analysis

1.   Question : Is there a difference in scores by gender?
2.   Analysis : Both of the bar charts show that the average scores and age by gender.
3.   Conclusion : There is no clear difference between age and score.
