```python
import pandas as pd
```

```python
data = {
    'name':['A','B','C','D'],
    'score':[80,90,75,88]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

```python
print("평균:", df['score'].mean())
```

```python
print(df[df['score']>=80])
```

df.head() --> 처음 5개 행 출력


df.info() --> 데이터프레임 정보 출력


df.describe() --> 기술통계량 요약 출력

```python
print(df[df['score']>=85])
```

```python
print(df.sort_values('name'))
```

```python
print(df.sort_values('score', ascending=False))
```
