```python
import pandas as pd
```

```python
data = {
    'name':['A','B','C','D'],
    'score':[80,90,75,88]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

```python
print("평균:", df['score'].mean())
```

```python
print(df[df['score']>=80])
```
