```python
import pandas as pd

data = {
    'movie': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'genre': ['Action', 'Drama', 'Action', 'Comedy', 'Drama', 'Comedy', 'Action', 'Drama', 'Comedy', 'Action'],
    'rating': [8.5, 7.2, 9.0, 6.5, 8.0, 7.0, 8.8, 7.5, 6.8, 9.2],
    'runtime': [120, 150, 110, 90, 140, 100, 130, 145, 95, 125]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

Question

*   Is there a difference in rating by genre?
*   Is there a relationship between runtime and rating?



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
df.groupby('genre')['rating'].mean().plot(kind='bar')
plt.title('Average Rating by Genre')

plt.subplot(1,3,2)
plt.scatter(df['runtime'], df['rating'])
plt.title('Runtime vs Rating')

plt.subplot(1,3,3)
df['rating'].plot(kind='hist')
plt.title('Distribution of rating')

plt.tight_layout()
plt.show()
```

Conclusion

*   Genre of action high in average rating.
*   There is no clear relationship between runtime and rating.
*   Rating distribute more between 7 and 9.
